```python

students = {
    101: {"name": "Vinayak", "branch": "CSE", "marks": (85, 90, 88)},
    102: {"name": "Ishaan", "branch": "ECE", "marks": (78, 82, 80)},
}


def add_student(roll_no, name, branch, marks_tuple):
    if roll_no in students:
        print(f"Student with Roll No {roll_no} already exists.")
    else:
        students[roll_no] = {"name": name, "branch": branch, "marks": marks_tuple}
        print(f"Student {name} added successfully.")


def delete_student(roll_no):
    if roll_no in students:
        removed = students.pop(roll_no)
        print(f"Student {removed['name']} (Roll No: {roll_no}) deleted.")
    else:
        print(f"Roll No {roll_no} not found.")


def update_student(roll_no, branch=None, marks_tuple=None):
    if roll_no in students:
        if branch:
            students[roll_no]["branch"] = branch
        if marks_tuple:
            students[roll_no]["marks"] = marks_tuple
        print(f"Record updated for Roll No {roll_no}.")
    else:
        print(f"Roll No {roll_no} not found.")


def display_students():
    print("\n--- Final Student Records ---")
    for roll_no, details in students.items():
        print(f"Roll No: {roll_no} | Name: {details['name']} | Branch: {details['branch']} | Marks: {details['marks']}")
    print("-----------------------------\n")


add_student(103, "Arijit", "ME", (91, 87, 85))
update_student(102, branch="Data Science", marks_tuple=(85, 88, 90))
delete_student(101)
display_students()
```
